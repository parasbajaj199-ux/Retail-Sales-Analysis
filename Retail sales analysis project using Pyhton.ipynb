# PHASE-> 7 Data Cleaning

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df["Year"] = df["Order_Date"].dt.year
```

```python
df["Month_Number"] = df["Order_Date"].dt.month
```

```python
df["Month_Name"] = df["Order_Date"].dt.month_name()
```

```python
df["Quarter"] = "Q" + df["Order_Date"].dt.quarter.astype(str)
```

```python
df[["Order_Date", "Year", "Month_Number", "Month_Name","Quarter"]].head()
```

```python
df.head()
```

# PHASE -> 8 Handle missing values

```python
df["Category"] = df["Category"].fillna("Undefine")
df["Region"] = df["Region"].fillna("Unkown")
df["Payment_Mode"] = df["Payment_Mode"].fillna("Not available")
```

```python
df["Payment_Mode"].unique()
```

```python
df["Region"].unique()
```

```python
df["Category"].unique()
```

```python
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].median())
```

```python
df["Sales"] = df["Sales"].fillna(df["Sales"].median())
```

```python
df["Profit"] = df["Profit"].fillna(df["Profit"].median())
```

```python
df[["Unit_Price","Sales","Profit"]].isnull().sum()
```

# PHASE -20-> Matplotlib 


```python
# Monthly sales trendS
df.head()
```

```python
monthly_sales = df.groupby(
    ["Month_Number","Month_Name"])["Sales"].sum().sort_index()
print(monthly_sales)
```

```python
#monthly_sales = df.groupby("Month_Name")["Sales"].sum()
monthly_sales = df.groupby(
    ["Month_Number","Month_Name"])["Sales"].sum().sort_index()
plt.figure(figsize= (15,7))
plt.plot(
   monthly_sales.index.get_level_values("Month_Name"), 
   monthly_sales.values,
   marker="o")
plt.title("Monthly sales trends")
plt.xlabel("Month Name")
plt.ylabel("Sales")
plt.show()
```

```python
# Sales Categeory -wise
sales_category = df.groupby("Category")["Sales"].sum()
plt.bar(sales_category.index,sales_category.values)
plt.title("Sales by  Category")
plt.xlabel("Sales Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
# Sales distribution
plt.hist(df["Sales"], bins=120)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```
